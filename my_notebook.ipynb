Reshaping : Reshape an input tensor in defined shape
Viewing : Work same as reshape but also change the original tensor
Stacking : combine multiple tensor on top of each tensor (vstack , hstack)
Squeeze : remove 1 dimension in the input tensor
unsqueeze : add 1 dimension to an input tensor


```python
# Reshaping , Viewing , Stacking
x = torch.arange(1,11)
print(x , x.shape)
print(x.reshape(2,5))
print(x)
print(x.view(1,10))

```

```python
# squeeze and unsqueeze
x = torch.tensor([[1,2,3,4,5]])
print(x , x.shape)
print(f"After Squuezing : {x.squeeze() , x.squeeze().shape}")
print(f"After Unsqueezing : {x.unsqueeze(dim=0) , x.unsqueeze(dim=1).shape}")
```

```python
# Permute
import torch
x = torch.rand(224,224,3)
print(f"Before Permutation : {x.shape}")
print(f"After Permutation : {x.permute(2,0,1).shape}")
```

# Reproducibility
in short how neural network learn
start with random number --> tensor operation --> update tensor number to try and make better representation for the data --> again --> again --> again

Essentially the random seed does is flavour the randomness

```python
import torch
tensor_A  = torch.rand(3,4)
tensor_B  = torch.rand(3,4)
print(tensor_A)
print(tensor_B)
print(tensor_A == tensor_B)
```

```python
import torch
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
tensor_A  = torch.rand(3,4)
torch.manual_seed(RANDOM_SEED)
tensor_B  = torch.rand(3,4)
print(tensor_A)
print(tensor_B)
print(tensor_A == tensor_B)
```
